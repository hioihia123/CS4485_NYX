# Feature Selection and Dataset Spliting

```python

import pandas as pd
import numpy as np
from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler

# Make a copy of the original dataset
data = df.copy()

# Convert the target to numeric
data["stress_score"] = pd.to_numeric(
    data["stress_score"], errors="coerce"
)

# Remove rows with invalid targets
data = data.dropna(subset=["stress_score"])
data = data[data["stress_score"] > 0].copy()

# Select numeric features, excluding identifiers and target
excluded_columns = {"id", "date", "stress_score"}

feature_columns = [
    col for col in data.select_dtypes(include=np.number).columns
    if col not in excluded_columns
    and data[col].notna().any()
    and data[col].nunique(dropna=True) > 1
]

# Remove rows with missing predictor values
data_complete = data.dropna(subset=feature_columns).copy()

print("Rows before removing missing features:", len(data))
print("Rows after removing missing features:", len(data_complete))
print("Rows removed:", len(data) - len(data_complete))

# Separate predictors, target, and participant IDs
X = data_complete[feature_columns].copy()
y = data_complete["stress_score"].copy()
groups = data_complete["id"].copy()

# Split by participant to avoid participant leakage
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx].copy()
X_test = X.iloc[test_idx].copy()

y_train = y.iloc[train_idx].copy()
y_test = y.iloc[test_idx].copy()

# Standardize using training data only
scaler = StandardScaler()

X_train = pd.DataFrame(
    scaler.fit_transform(X_train),
    columns=feature_columns,
    index=X_train.index
)

X_test = pd.DataFrame(
    scaler.transform(X_test),
    columns=feature_columns,
    index=X_test.index
)

print("\nTraining rows:", len(X_train))
print("Testing rows:", len(X_test))
print(
    "Training participants:",
    groups.iloc[train_idx].nunique()
)
print(
    "Testing participants:",
    groups.iloc[test_idx].nunique()
)
print(
    "Participants shared between sets:",
    len(
        set(groups.iloc[train_idx])
        & set(groups.iloc[test_idx])
    )
)

print("\nFinal training features:")
display(X_train.head())
```

```python

from sklearn.model_selection import GroupKFold
from xgboost import XGBRegressor

# Participant IDs aligned with the stress-model training rows
train_groups = data_complete.loc[
    X_train_original.index, "id"
].astype(str)

# Generate out-of-fold stress predictions for training rows
oof_stress_predictions = pd.Series(
    index=X_train_original.index,
    dtype=float
)

group_count = train_groups.nunique()
n_splits = min(5, group_count)

if n_splits < 2:
    raise ValueError("Not enough training participants for cross-validation.")

group_cv = GroupKFold(n_splits=n_splits)

for fold, (fit_idx, val_idx) in enumerate(
    group_cv.split(
        X_train_original,
        y_train,
        groups=train_groups
    ),
    start=1
):
    fold_model = XGBRegressor(
        n_estimators=100,
        max_depth=3,
        learning_rate=0.1,
        objective="reg:squarederror",
        random_state=42
    )

    # Train on participants assigned to this fold's training portion
    fold_model.fit(
        X_train_original.iloc[fit_idx],
        y_train.iloc[fit_idx]
    )

    # Predict participants the fold model has not seen
    oof_stress_predictions.iloc[val_idx] = (
        fold_model.predict(X_train_original.iloc[val_idx])
    )

# Store out-of-fold predictions for training participants
train_stress = data_complete.loc[
    X_train_original.index, ["id", "date", "cutoff_hour"]
].copy()

train_stress["id"] = train_stress["id"].astype(str)
train_stress["date"] = pd.to_datetime(
    train_stress["date"], errors="coerce"
).dt.normalize()

train_stress["predicted_stress_score"] = (
    oof_stress_predictions
)

# Store predictions for held-out test participants
test_stress = data_complete.loc[
    X_test_original.index, ["id", "date", "cutoff_hour"]
].copy()

test_stress["id"] = test_stress["id"].astype(str)
test_stress["date"] = pd.to_datetime(
    test_stress["date"], errors="coerce"
).dt.normalize()

test_stress["predicted_stress_score"] = y_pred_xgb

# Combine the training and test predictions
stress_predictions = pd.concat(
    [train_stress, test_stress],
    ignore_index=True
)

print("Stress predictions created!")
print("Rows:", len(stress_predictions))
print("Missing predictions:",
      stress_predictions["predicted_stress_score"].isna().sum())

display(stress_predictions.head())
```

```python
# Keep only the 9 PM stress predictions
stress_9pm = stress_predictions[
    stress_predictions["cutoff_hour"] == 21
].copy()

# Keep only the columns needed from the stress model
stress_9pm = stress_9pm[
    ["id", "date", "predicted_stress_score"]
]

# Check for duplicate participant-date records
print(
    "Duplicate 9 PM participant-date records:",
    stress_9pm.duplicated(["id", "date"]).sum()
)

# Make sure daily data uses matching key formats
daily_sleep = daily.copy()

daily_sleep["id"] = daily_sleep["id"].astype(str)
daily_sleep["date"] = pd.to_datetime(
    daily_sleep["date"], errors="coerce"
).dt.normalize()

# Convert sleep efficiency to numeric
if "sleep_efficiency" not in daily_sleep.columns:
    raise ValueError(
        "sleep_efficiency was not found. "
        "Check the sleep-related columns printed in Step 2."
    )

daily_sleep["sleep_efficiency"] = pd.to_numeric(
    daily_sleep["sleep_efficiency"], errors="coerce"
)

# Keep the sleep target and keys
sleep_df = daily_sleep[
    ["id", "date", "sleep_efficiency"]
].copy()

# Add the previous night's sleep efficiency.
# The shift occurs within each participant, in date order.
daily_sorted = daily_sleep.sort_values(
    ["id", "date"]
).copy()

daily_sorted["previous_sleep_efficiency"] = (
    daily_sorted.groupby("id")["sleep_efficiency"].shift(1)
)

previous_sleep = daily_sorted[
    ["id", "date", "previous_sleep_efficiency"]
]

# Merge the previous-night feature into the target dataframe
sleep_df = sleep_df.merge(
    previous_sleep,
    on=["id", "date"],
    how="left",
    validate="one_to_one"
)

# Add the day of the week as an optional feature
sleep_df["day_of_week"] = sleep_df["date"].dt.dayofweek

# Merge in predicted stress score
sleep_df = sleep_df.merge(
    stress_9pm,
    on=["id", "date"],
    how="inner",
    validate="one_to_one"
)

# Remove rows without a target or a stress prediction
sleep_df = sleep_df.dropna(
    subset=["sleep_efficiency", "predicted_stress_score"]
).copy()

# Sort for easier inspection
sleep_df = sleep_df.sort_values(
    ["id", "date"]
).reset_index(drop=True)

print("Sleep-model dataframe created!")
print("Rows:", len(sleep_df))
print("Columns:", len(sleep_df.columns))
print("Participants:", sleep_df["id"].nunique())

print("\nFinal columns:")
print(sleep_df.columns.tolist())

display(sleep_df.head(10))
```

```python
# Get participant IDs from the original stress-model split
stress_train_ids = set(
    data_complete.loc[X_train_original.index, "id"].astype(str)
)

stress_test_ids = set(
    data_complete.loc[X_test_original.index, "id"].astype(str)
)

# Verify that the original stress-model groups do not overlap
print("STRESS MODEL PARTICIPANT SPLIT")
print("Training participants:", len(stress_train_ids))
print("Testing participants:", len(stress_test_ids))
print(
    "Participants shared between splits:",
    len(stress_train_ids.intersection(stress_test_ids))
)

# Get participants present in the sleep-model dataframe
sleep_ids = set(sleep_df["id"].astype(str))

# Determine which original participants appear in the sleep data
sleep_train_ids = sleep_ids.intersection(stress_train_ids)
sleep_test_ids = sleep_ids.intersection(stress_test_ids)

print("\nSLEEP MODEL PARTICIPANT SPLIT")
print("Training participants present:", len(sleep_train_ids))
print("Testing participants present:", len(sleep_test_ids))
print(
    "Participants shared between sleep splits:",
    len(sleep_train_ids.intersection(sleep_test_ids))
)

# Check whether any sleep participant is absent from the stress split
unassigned_ids = sleep_ids - stress_train_ids - stress_test_ids

print("\nParticipants without an original split:", len(unassigned_ids))

# Display a clear pass/fail result
if (
    stress_train_ids.isdisjoint(stress_test_ids)
    and not unassigned_ids
    and sleep_train_ids.isdisjoint(sleep_test_ids)
):
    print("\nPASS: Sleep participants follow the original stress-model split.")
else:
    print("\nWARNING: Investigate the participant split before modeling.")
```

```python
# Use the cleaned daily sleep dataset
daily_check = daily.copy()

daily_check["id"] = daily_check["id"].astype(str)
daily_check["date"] = pd.to_datetime(
    daily_check["date"], errors="coerce"
).dt.normalize()

daily_check["sleep_efficiency"] = pd.to_numeric(
    daily_check["sleep_efficiency"], errors="coerce"
)

# Check that there is only one row per participant and date
duplicate_count = daily_check.duplicated(
    subset=["id", "date"]
).sum()

print("Duplicate participant-date records:", duplicate_count)

if duplicate_count > 0:
    raise ValueError(
        "Duplicate participant-date records found. "
        "Resolve these before checking previous-night alignment."
    )

# Create a lookup table containing each participant's sleep efficiency
previous_lookup = daily_check[
    ["id", "date", "sleep_efficiency"]
].copy()

# Move each source date forward one day:
# Monday's sleep efficiency becomes available as Tuesday's prior-night feature
previous_lookup["date"] = (
    previous_lookup["date"] + pd.Timedelta(days=1)
)

previous_lookup = previous_lookup.rename(
    columns={
        "sleep_efficiency": "expected_previous_sleep_efficiency"
    }
)

# Match each sleep date to the exact previous calendar date
alignment_check = sleep_df[
    ["id", "date", "previous_sleep_efficiency"]
].copy()

alignment_check["id"] = alignment_check["id"].astype(str)
alignment_check["date"] = pd.to_datetime(
    alignment_check["date"], errors="coerce"
).dt.normalize()

alignment_check = alignment_check.merge(
    previous_lookup,
    on=["id", "date"],
    how="left",
    validate="one_to_one"
)

# Compare existing and expected previous-night values
alignment_check["matches_expected"] = np.isclose(
    alignment_check["previous_sleep_efficiency"],
    alignment_check["expected_previous_sleep_efficiency"],
    equal_nan=True
)

print("\nPREVIOUS-NIGHT ALIGNMENT RESULTS")
print("Sleep rows checked:", len(alignment_check))
print(
    "Rows matching exact prior calendar date:",
    int(alignment_check["matches_expected"].sum())
)
print(
    "Rows not matching:",
    int((~alignment_check["matches_expected"]).sum())
)

if alignment_check["matches_expected"].all():
    print("\nPASS: Previous-night values align with calendar dates.")
else:
    print("\nWARNING: Some previous-night values are misaligned.")
    display(
        alignment_check.loc[
            ~alignment_check["matches_expected"]
        ].head(20)
    )
```

```python
# Choose the initial sleep-model features
sleep_features = [
    "predicted_stress_score",
    "previous_sleep_efficiency",
    "day_of_week"
]

sleep_target = "sleep_efficiency"

# Confirm that all required columns exist
required_columns = sleep_features + [
    sleep_target, "id", "date"
]

missing_columns = [
    col for col in required_columns
    if col not in sleep_df.columns
]

if missing_columns:
    raise ValueError(
        "Missing required columns: " + ", ".join(missing_columns)
    )

# Separate participants according to the original stress-model split
sleep_train = sleep_df[
    sleep_df["id"].astype(str).isin(stress_train_ids)
].copy()

sleep_test = sleep_df[
    sleep_df["id"].astype(str).isin(stress_test_ids)
].copy()

# Keep rows with a known sleep target and stress prediction.
# Missing previous-night sleep values will be imputed using training data only.
sleep_train = sleep_train.dropna(
    subset=[sleep_target, "predicted_stress_score"]
)

sleep_test = sleep_test.dropna(
    subset=[sleep_target, "predicted_stress_score"]
)

# Create X and y
X_sleep_train = sleep_train[sleep_features].copy()
y_sleep_train = sleep_train[sleep_target].copy()

X_sleep_test = sleep_test[sleep_features].copy()
y_sleep_test = sleep_test[sleep_target].copy()

print("Sleep-model training rows:", len(X_sleep_train))
print("Sleep-model testing rows:", len(X_sleep_test))

print(
    "Training participants:",
    sleep_train["id"].nunique()
)
print(
    "Testing participants:",
    sleep_test["id"].nunique()
)

print(
    "Participants shared:",
    len(
        set(sleep_train["id"])
        .intersection(set(sleep_test["id"]))
    )
)

print("\nTraining features:")
display(X_sleep_train.head())

print("\nTesting features:")
display(X_sleep_test.head())
```

```python

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Fill missing predictor values using training-set medians
sleep_imputer = SimpleImputer(strategy="median")

X_sleep_train_imputed = sleep_imputer.fit_transform(
    X_sleep_train
)

X_sleep_test_imputed = sleep_imputer.transform(
    X_sleep_test
)

# Standardize for models such as Linear Regression, Ridge, or SVR
sleep_scaler = StandardScaler()

X_sleep_train_scaled = sleep_scaler.fit_transform(
    X_sleep_train_imputed
)

X_sleep_test_scaled = sleep_scaler.transform(
    X_sleep_test_imputed
)

# Convert to DataFrames for easier inspection
X_sleep_train_scaled = pd.DataFrame(
    X_sleep_train_scaled,
    columns=sleep_features,
    index=X_sleep_train.index
)

X_sleep_test_scaled = pd.DataFrame(
    X_sleep_test_scaled,
    columns=sleep_features,
    index=X_sleep_test.index
)

print("Sleep-model preprocessing complete.")
print("Training shape:", X_sleep_train_scaled.shape)
print("Testing shape:", X_sleep_test_scaled.shape)
print("Missing values in training features:",
      int(X_sleep_train_scaled.isna().sum().sum()))
print("Missing values in testing features:",
      int(X_sleep_test_scaled.isna().sum().sum()))
```

```python
# Define the features for each modality

modality_features = {
    "heart_rate_input": [
        "bpm",
        "bpm_std",
        "bpm_max"
    ],

    "activity_input": [
        "steps",
        "steps_std",
        "calories",
        "distance",
        "active_hours"
    ],

    "temperature_input": [
        "temperature"
    ],

    "context_input": [
        "previous_sleep_efficiency",
        "observed_hours",
        "latest_hour",
        "cutoff_hour",
        "hour_coverage"
    ]
}

# Keep only features that actually exist in the dataset
for modality in modality_features:
    modality_features[modality] = [
        col for col in modality_features[modality]
        if col in X_train_original.columns
    ]

# Remove empty branches
modality_features = {
    name: columns
    for name, columns in modality_features.items()
    if columns
}

print("Features used by each modality:")
for modality, columns in modality_features.items():
    print(f"\n{modality}: {columns}")

print("\nNumber of modality branches:", len(modality_features))
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler

# Get participant IDs in the same row order as X_train_original
train_groups = (
    data_complete.loc[X_train_original.index, "id"]
    .astype(str)
    .to_numpy()
)

# Split training data into fitting and validation subsets
group_split = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15,
    random_state=42
)

fit_idx, val_idx = next(
    group_split.split(X_train_original, y_train, groups=train_groups)
)

X_fit = X_train_original.iloc[fit_idx].copy()
X_val = X_train_original.iloc[val_idx].copy()

y_fit = y_train.iloc[fit_idx].copy()
y_val = y_train.iloc[val_idx].copy()

fit_ids = set(train_groups[fit_idx])
val_ids = set(train_groups[val_idx])

print("Model-fitting rows:", len(X_fit))
print("Validation rows:", len(X_val))
print("Model-fitting participants:", len(fit_ids))
print("Validation participants:", len(val_ids))
print("Participant overlap:", len(fit_ids & val_ids))
```

```python
import tensorflow as tf

from tensorflow.keras import Model
from tensorflow.keras.layers import (
    Input, Dense, Concatenate, Dropout
)
from tensorflow.keras.callbacks import EarlyStopping

# Set random seeds for more reproducible results
np.random.seed(42)
tf.keras.utils.set_random_seed(42)


def build_fusion_model(modality_features):

    inputs = {}
    branch_outputs = []

    # Create one separate branch for each modality
    for input_name, columns in modality_features.items():

        modality_input = Input(
            shape=(len(columns),),
            name=input_name
        )

        # Each modality learns its own representation
        branch = Dense(8, activation="relu")(modality_input)
        branch = Dense(4, activation="relu")(branch)

        inputs[input_name] = modality_input
        branch_outputs.append(branch)

    # Combine all modality representations
    if len(branch_outputs) > 1:
        combined = Concatenate()(branch_outputs)
    else:
        combined = branch_outputs[0]

    # Learn relationships across modalities
    combined = Dense(16, activation="relu")(combined)
    combined = Dropout(0.2)(combined)
    combined = Dense(8, activation="relu")(combined)

    # Predict one continuous stress score
    output = Dense(1, activation="linear", name="stress_score")(combined)

    model = Model(inputs=inputs, outputs=output)

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss="mse",
        metrics=["mae"]
    )

    return model


fusion_model = build_fusion_model(modality_features)

fusion_model.summary()
```

```python
# Fit the scaler using only the model-fitting rows
fusion_scaler = StandardScaler()

X_fit_scaled = pd.DataFrame(
    fusion_scaler.fit_transform(X_fit),
    columns=X_fit.columns,
    index=X_fit.index
)

# Transform validation rows without refitting
X_val_scaled = pd.DataFrame(
    fusion_scaler.transform(X_val),
    columns=X_val.columns,
    index=X_val.index
)


# Convert the feature DataFrame into separate model inputs
def make_model_inputs(X, modality_features):

    model_inputs = {}

    for input_name, columns in modality_features.items():
        model_inputs[input_name] = X[columns].to_numpy(
            dtype=np.float32
        )

    return model_inputs


fit_inputs = make_model_inputs(
    X_fit_scaled, modality_features
)

val_inputs = make_model_inputs(
    X_val_scaled, modality_features
)

print("Training inputs prepared for", len(fit_inputs), "branches.")
```

```python
# Create a fresh model
fusion_model = build_fusion_model(modality_features)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=15,
    restore_best_weights=True
)

history = fusion_model.fit(
    fit_inputs,
    y_fit.to_numpy(dtype=np.float32),
    validation_data=(
        val_inputs,
        y_val.to_numpy(dtype=np.float32)
    ),
    epochs=150,
    batch_size=32,
    callbacks=[early_stop],
    verbose=1
)

best_epoch = np.argmin(history.history["val_loss"]) + 1

print("\nBest epoch:", best_epoch)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Fit a new scaler on the complete training set
final_scaler = StandardScaler()

X_train_fusion_scaled = pd.DataFrame(
    final_scaler.fit_transform(X_train_original),
    columns=X_train_original.columns,
    index=X_train_original.index
)

X_test_fusion_scaled = pd.DataFrame(
    final_scaler.transform(X_test_original),
    columns=X_test_original.columns,
    index=X_test_original.index
)

# Create separate inputs for each modality
X_train_fusion_inputs = make_model_inputs(
    X_train_fusion_scaled, modality_features
)

X_test_fusion_inputs = make_model_inputs(
    X_test_fusion_scaled, modality_features
)

# Train a fresh model on the complete training set
final_fusion_model = build_fusion_model(modality_features)

final_fusion_model.fit(
    X_train_fusion_inputs,
    y_train.to_numpy(dtype=np.float32),
    epochs=best_epoch,
    batch_size=32,
    verbose=0
)

# Predict scores for held-out participants
y_pred_fusion = final_fusion_model.predict(
    X_test_fusion_inputs,
    verbose=0
).flatten()

# Evaluate predictions
fusion_mae = mean_absolute_error(y_test, y_pred_fusion)
fusion_rmse = np.sqrt(
    mean_squared_error(y_test, y_pred_fusion)
)
fusion_r2 = r2_score(y_test, y_pred_fusion)

print("Multimodal Fusion Results")
print("-------------------------")
print(f"MAE:  {fusion_mae:.4f}")
print(f"RMSE: {fusion_rmse:.4f}")
print(f"R²:   {fusion_r2:.4f}")
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import GroupKFold
from sklearn.preprocessing import StandardScaler
from sklearn.base import clone

import tensorflow as tf

# Get participant IDs aligned with the stress training rows
train_groups = (
    data_complete.loc[X_train_original.index, "id"]
    .astype(str)
    .to_numpy()
)

# Get participant IDs aligned with the stress test rows
test_groups = (
    data_complete.loc[X_test_original.index, "id"]
    .astype(str)
    .to_numpy()
)

# Convert a feature dataframe into inputs for the fusion model
def prepare_fusion_inputs(X, scaler, modality_features):
    X_scaled = pd.DataFrame(
        scaler.transform(X),
        columns=X.columns,
        index=X.index
    )

    return {
        name: X_scaled[columns].to_numpy(dtype=np.float32)
        for name, columns in modality_features.items()
    }


# Generate out-of-fold predictions for stress training participants
oof_stress_predictions = pd.Series(
    index=X_train_original.index,
    dtype=float
)

group_cv = GroupKFold(n_splits=5)

for fold, (fit_idx, val_idx) in enumerate(
    group_cv.split(X_train_original, y_train, groups=train_groups),
    start=1
):
    X_fold_train = X_train_original.iloc[fit_idx]
    X_fold_val = X_train_original.iloc[val_idx]

    y_fold_train = y_train.iloc[fit_idx]

    # Scale using only this fold's training rows
    fold_scaler = StandardScaler()
    fold_scaler.fit(X_fold_train)

    fold_train_inputs = prepare_fusion_inputs(
        X_fold_train, fold_scaler, modality_features
    )
    fold_val_inputs = prepare_fusion_inputs(
        X_fold_val, fold_scaler, modality_features
    )

    # Create a new model for this fold
    tf.keras.utils.set_random_seed(42 + fold)

    fold_model = build_fusion_model(modality_features)

    fold_model.fit(
        fold_train_inputs,
        y_fold_train.to_numpy(dtype=np.float32),
        epochs=best_epoch,
        batch_size=32,
        verbose=0
    )

    fold_predictions = fold_model.predict(
        fold_val_inputs,
        verbose=0
    ).flatten()

    oof_stress_predictions.iloc[val_idx] = fold_predictions

    print(f"Finished fold {fold}/5")

print("\nTraining out-of-fold predictions complete.")
```
